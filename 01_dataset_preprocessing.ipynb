```python
from pathlib import Path
import uuid

# 데이터셋 상위 폴더 경로 (본인 경로로 수정)
DATASET_DIR = Path("/content/drive/MyDrive/bus_dataset")

# 이미지가 들어 있는 폴더
FOLDERS = ["0", "1", "2plus", "N"]

# 이미지 확장자
IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".webp"}

for folder_name in FOLDERS:
    folder_path = DATASET_DIR / folder_name

    if not folder_path.is_dir():
        print(f"폴더 없음: {folder_path}")
        continue

    # 이미지 파일만 가져오기
    images = sorted(
        [
            file for file in folder_path.iterdir()
            if file.is_file() and file.suffix.lower() in IMAGE_EXTENSIONS
        ],
        key=lambda file: file.name.lower()
    )

    # 기존 파일명 충돌 방지를 위해 임시 이름으로 변경
    temp_files = []

    for image in images:
        temp_path = folder_path / f"__temp_{uuid.uuid4().hex}{image.suffix.lower()}"
        image.rename(temp_path)
        temp_files.append(temp_path)

    # 폴더별로 001부터 번호 부여
    for index, image in enumerate(temp_files, start=1):
        new_name = f"{folder_name}_{index:03d}{image.suffix.lower()}"
        image.rename(folder_path / new_name)

    print(f"{folder_name} 폴더: {len(images)}장 변경 완료")

print("\n전체 파일명 정리 완료!")
```

```python
from pathlib import Path
from collections import Counter

# 기존 데이터셋 경로
DATASET_DIR = Path("/content/drive/MyDrive/bus_dataset")

FOLDERS = ["0", "1", "2plus", "N"]

print("데이터셋 경로:", DATASET_DIR)
print("경로 존재 여부:", DATASET_DIR.exists())

for folder_name in FOLDERS:
    folder_path = DATASET_DIR / folder_name

    print(f"\n===== {folder_name} 폴더 =====")

    if not folder_path.is_dir():
        print("폴더를 찾을 수 없습니다.")
        continue

    files = [f for f in folder_path.iterdir() if f.is_file()]
    subfolders = [f for f in folder_path.iterdir() if f.is_dir()]

    print("전체 파일 수:", len(files))
    print("하위 폴더 수:", len(subfolders))
    print("확장자 종류:", dict(Counter(f.suffix.lower() for f in files)))
    print("파일명 예시:", [f.name for f in files[:5]])
    print("하위 폴더:", [f.name for f in subfolders[:10]])
```

```python
from pathlib import Path
from collections import Counter

# Google Drive 내에서 데이터셋 폴더 찾기
DRIVE_DIR = Path("/content/drive/MyDrive/bus_dataset")

FOLDERS = ["0", "1", "2plus", "N"]
IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".webp", ".avif", ".jfif"}

print("===== 이미지 폴더 검색 =====")

for folder in DRIVE_DIR.rglob("*"):
    if folder.is_dir() and folder.name in FOLDERS:

        images = [
            f for f in folder.iterdir()
            if f.is_file() and f.suffix.lower() in IMAGE_EXTENSIONS
        ]

        if images:
            print(f"\n폴더: {folder}")
            print(f"이미지 수: {len(images)}장")
            print("확장자:", dict(Counter(f.suffix.lower() for f in images)))

print("\n검색 완료!")
```
