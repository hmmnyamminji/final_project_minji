```python
from pathlib import Path
import uuid

# 데이터셋 상위 폴더 경로 (본인 경로로 수정)
DATASET_DIR = Path("/content/drive/MyDrive/bus_dataset/N1")

# 이미지가 들어 있는 폴더
FOLDERS = ["N1"]

# 이미지 확장자
IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".webp"}

for folder_name in FOLDERS:
    folder_path = DATASET_DIR / folder_name

    if not folder_path.is_dir():
        print(f"폴더 없음: {folder_path}")
        continue

    # 이미지 파일만 가져오기
    images = sorted(
        [
            file for file in folder_path.iterdir()
            if file.is_file() and file.suffix.lower() in IMAGE_EXTENSIONS
        ],
        key=lambda file: file.name.lower()
    )

    # 기존 파일명 충돌 방지를 위해 임시 이름으로 변경
    temp_files = []

    for image in images:
        temp_path = folder_path / f"__temp_{uuid.uuid4().hex}{image.suffix.lower()}"
        image.rename(temp_path)
        temp_files.append(temp_path)

    # 폴더별로 001부터 번호 부여
    for index, image in enumerate(temp_files, start=1):
        new_name = f"{folder_name}_{index:03d}{image.suffix.lower()}"
        image.rename(folder_path / new_name)

    print(f"{folder_name} 폴더: {len(images)}장 변경 완료")

print("\n전체 파일명 정리 완료!")
```
